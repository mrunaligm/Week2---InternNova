## Task 3 — Data Cleaning (20 marks)
The raw dataset contains missing values, one duplicate row, and a numeric value stored as text.

```python
df = pd.read_csv("student_performance_raw.csv")
numeric_cols = ["Maths", "Science", "English", "Attendance"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
print("Missing values before cleaning:\n", df.isna().sum())
print("Duplicate rows before cleaning:", df.duplicated().sum())
df = df.drop_duplicates().copy()
for col in ["Maths", "Science", "English"]:
    df[col] = df[col].fillna(df[col].mean())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())
print("Missing values after cleaning:\n", df.isna().sum())
print("Duplicate rows after cleaning:", df.duplicated().sum())
display(df.head())
```

## Task 4 — Data Visualization (20 marks)

```python
import matplotlib.pyplot as plt
plot_df = pd.read_csv("student_performance_cleaned.csv")
plt.figure(figsize=(9,4.5))
plt.bar(plot_df["Name"].head(10), plot_df["Average_Marks"].head(10))
plt.title("Average Marks of First 10 Students")
plt.xlabel("Student"); plt.ylabel("Average marks")
plt.xticks(rotation=35, ha="right"); plt.tight_layout(); plt.show()
```

**Bar chart:** Compares the average marks of the first ten students to make higher and lower scores easy to spot.

```python
plt.figure(figsize=(9,4.5))
plt.plot(plot_df["Student_ID"], plot_df["Average_Marks"], marker="o")
plt.title("Student Average Marks by ID")
plt.xlabel("Student ID"); plt.ylabel("Average marks")
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

**Line chart:** Shows how average marks vary across student IDs. The connecting line is for comparison, not a time trend.

```python
plt.figure(figsize=(7,4.5))
plt.hist(plot_df["Average_Marks"], bins=6, edgecolor="black")
plt.title("Distribution of Student Average Marks")
plt.xlabel("Average marks"); plt.ylabel("Number of students")
plt.tight_layout(); plt.show()
```

**Histogram:** Groups average marks into ranges and shows how many students fall in each range.

```python
plt.figure(figsize=(7,4.5))
plt.scatter(plot_df["Attendance"], plot_df["Average_Marks"])
plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance (%)"); plt.ylabel("Average marks")
plt.tight_layout(); plt.show()
```

**Scatter plot:** Explores whether attendance and average marks appear related; correlation does not prove causation.

## Task 5 — Mini Project: Student Performance Data Analysis (20 marks)

```python
df = pd.read_csv("student_performance_cleaned.csv")
subjects = ["Maths", "Science", "English"]
print("Dataset shape:", df.shape)
print("Important columns:", ["Name"] + subjects + ["Attendance", "Average_Marks", "Result"])
display(df.head())
df.info()
print("Average subject score:", round(df[subjects].to_numpy().mean(), 2))
print("Highest subject mark:", df[subjects].max().max())
print("Lowest subject mark:", df[subjects].min().min())
print("Students passed:", int((df["Result"] == "Pass").sum()))
print("Students failed:", int((df["Result"] == "Fail").sum()))
print("Average by subject:\n", df[subjects].mean().round(2))
```

### Three mini-project charts

```python
counts = df["Result"].value_counts().reindex(["Pass","Fail"], fill_value=0)
plt.figure(figsize=(6,4)); plt.bar(subjects, df[subjects].mean())
plt.title("Average Marks by Subject"); plt.ylabel("Average marks"); plt.tight_layout(); plt.show()
plt.figure(figsize=(6,4)); plt.bar(counts.index, counts.values)
plt.title("Pass and Fail Counts"); plt.ylabel("Number of students"); plt.tight_layout(); plt.show()
plt.figure(figsize=(6,4)); plt.scatter(df["Attendance"], df["Average_Marks"])
plt.title("Attendance vs Average Marks"); plt.xlabel("Attendance (%)"); plt.ylabel("Average marks"); plt.tight_layout(); plt.show()
```
